# GHSOM training on Iris dataset

Different variants of GHSOM are trained on the Iris dataset to be compared with the standard SOM in the thesis

```python
import numpy as np
import pandas as pd
from GHSOM import GHSOM
from help_functions import *
from sklearn.preprocessing import MinMaxScaler
```

Load and preprocess the data

```python
df = pd.read_csv("../data/iris.csv")
y = df['Iris-setosa'].to_numpy()
X = df.drop(['Iris-setosa'], axis=1)

# Normalize dataset
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
```

## Initialize deep GHSOM

Details:
- Input dimension of size 4
- Tau1 parameter of size 0.8
- Tau2 parameter of size 0.0035
- 30 training epoch for each map
- Initial learning rate of 0.5
- Beta of size 0.999
- QE metric used for vertical expansion
- Minimum of 5 samples for vertical expansion
- Maximum map dimension of 30 for individual GSOM
- Euclidean distance used to find neigbors
- Gaussian neigborhood function used
- Exponential decay function used

```python
ghsom_deep = GHSOM(
    input_dim=4,
    t1=0.8,
    t2=0.0035,
    training_epoch_num=30)
```

```python
ghsom_deep.train(X, y)
```

Plot hierarchical label matrix

```python
plot_ghsom(ghsom_deep, X, y)
```

Number of nodes used to represent the data

```python
len(ghsom_deep.terminal_units)
```
